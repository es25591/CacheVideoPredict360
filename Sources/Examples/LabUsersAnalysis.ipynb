```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Load the CSV file
df = pd.read_csv('active_users_per_step.csv')

print("\nFirst few rows:")
print(df.head(10))
```

```python
#plot histogram of active users per step
plt.figure(figsize=(10, 6))
plt.hist(df['num_active_users'], bins=30, color='skyblue', edgecolor='black')
plt.title('Histogram of Active Users per Step')
plt.xlabel('Number of Active Users')
plt.ylabel('Frequency')
plt.grid(axis='y', alpha=0.75)
plt.show()
```
