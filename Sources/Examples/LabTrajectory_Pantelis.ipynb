```python
import os
import numpy as np
import matplotlib.pyplot as plt
```

```python
def load_trajectories(foulder_path):
    videos_trajectories = {}

    files = os.listdir(foulder_path)
    for file in files:
        idx = int(file[1:]) - 1
        videos_trajectories[idx] = []
        path = os.path.join(foulder_path, file)
        
        videos_files = os.listdir(path)
        for trajectory in videos_files:
            trajectory_path = os.path.join(path, trajectory)
            data = np.loadtxt(trajectory_path, delimiter=',')
            videos_trajectories[idx].append(data)
            
    
    return videos_trajectories
```

```python
def map_tiles_trajectory(id, n=4, m=3):
    if id == 1:
        tiles = [0,1,2,3]
    elif id == 2:
        tiles = [0,1,4,5]
    elif id == 3:
        tiles = [1,2,5,6]
    elif id == 4:
        tiles = [2,3,6,7]
    elif id == 5:
        tiles = [0,4,3,7]
    elif id == 6:
        tiles = [4,5,8,9]
    elif id == 7:
        tiles = [5,6,9,10]
    elif id == 8:
        tiles = [6,7,10,11]
    elif id == 9:
        tiles = [4,8,7,11]
    else:
        tiles = [8,9,10,11]
    
    frame = np.zeros(n * m)
    frame[tiles] = 1

    return frame
```

```python
def simulate_viewport(video, trajectories, n=4, m=3):
    video %= 10
    trajectories = trajectories[video]
    
    trajectory = trajectories[np.random.randint(len(trajectories))]
    videos_viewport_tiles = [
        map_tiles_trajectory(int(tile_id)) for tile_id in trajectory
    ]
    
    user_viewport = []

    for tile in videos_viewport_tiles:
        viewport = set()
        for i in range(n):
            for j in range(m):
                if tile[i * m + j] == 1:
                    viewport.add((i, j))
        user_viewport.append(viewport)
    
    return user_viewport
```
